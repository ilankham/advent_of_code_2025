# Set puzzle parameters and create AoC Session

```python
# set puzzle parameters
PUZZLE_DAY = 8
PUZZLE_YEAR = 2025

# import from local packages
from aoc_solver import AoCSession, AoCSolver, AoCTester
AoC_SESSION = AoCSession.from_file()
```

# Import additional packages

```python
# import from standard library packages
from  math import inf, sqrt
from typing import List, Set, Tuple, Union
```

# Create solver class and instance

```python
class Solver(AoCSolver):

    @staticmethod
    def build_circuits(
        data: Tuple[str],
        number_of_connections:Union[int,float]=10
    ) -> Tuple[List[Set[Tuple[int,int,int]]], Set[Tuple[int,int,int]]]:
        number_of_points = len(data)
        distances = {}
        for i in range(number_of_points):
            for j in range(i+1,number_of_points):
                point1 = tuple(int(n) for n in data[i].split(','))
                point2 = tuple(int(n) for n in data[j].split(','))
                distances[(point1, point2)] = sqrt(
                    (point1[0] - point2[0])**2 +
                    (point1[1] - point2[1])**2 +
                    (point1[2] - point2[2])**2
                )
        distances_sorted = dict(sorted(distances.items(), key=lambda item: item[1]))
        circuits = []
        for n, (connection, distance) in enumerate(distances_sorted.items()):
            if n >= number_of_connections:
                break
            point1 = connection[0]
            point2 = connection[1]
            i = 0
            new_circuit = []
            point1_found = False
            point2_found = False
            while i < len(circuits):
                if point1 in circuits[i]:
                    point1_found = True
                if point2 in circuits[i]:
                    point2_found = True
                if point1_found or point2_found:
                    new_circuit.extend({point1, point2})
                    new_circuit.extend(circuits[i])
                    del circuits[i]
                    continue
                i += 1
            if new_circuit:
                circuits.append(set(new_circuit))
            else:
                circuits.append({point1, point2})
            if not(point1_found and point2_found):
                latest_connection = {point1, point2}
        return circuits, latest_connection

    def solve_part1(self, data: Tuple[str], number_of_connections=10) -> int:
        circuits, _ = self.build_circuits(data, number_of_connections)
        circuits.sort(key=lambda x: len(x), reverse=True)
        print(f'{circuits=}')
        return len(circuits[0])*len(circuits[1])*len(circuits[2])

    def solve_part2(self, data: Tuple[str]) -> int:
        _, lastest_connection = self.build_circuits(data, number_of_connections=inf)
        return_value = 1
        for point in lastest_connection:
            return_value *= point[0]
        return return_value
```

```python
solver = Solver(PUZZLE_YEAR, PUZZLE_DAY, AoC_SESSION)
```

# Build part 1 test case(s)

```python
puzzle_instructions = solver.puzzle_instructions

part1_test_input = solver.get_value_after('For example:').create_tuple(str, '\n')
print(f'{part1_test_input=}\n')

part1_test_output = solver.get_value_after(' produces ').as_int
print(f'{part1_test_output=}\n')
```

```python
part_1_tester = AoCTester()
part_1_tester.add_test_case(part1_test_input, part1_test_output)
```

```python
%%time
part_1_tester.run_tests(solver.solve_part1, number_of_connections=10)
```

# Determine part 1 solution

```python
%%time
puzzle_input = solver.puzzle_input.create_tuple(str, '\n')
part1_solution = solver.solve_part1(puzzle_input, number_of_connections=1000)
print(f'{part1_solution=}\n')
```
