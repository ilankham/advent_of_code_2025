# Set puzzle parameters and create AoC Session

```python
# set puzzle parameters
PUZZLE_DAY = 5
PUZZLE_YEAR = 2025

# import from local packages
from aoc_solver import AoCSession, AoCSolver, AoCTester
AoC_SESSION = AoCSession.from_file()
```

# Import additional packages

```python
# import from standard library packages
from typing import Tuple

# import from third-party packages
from tqdm.notebook import tqdm
```

# Create solver class and instance

```python
class Solver(AoCSolver):
    def solve_part1(self, data: Tuple[str,str]) -> int:
        fresh_ID_ranges = [
            tuple(int(endpoint) for endpoint in ID_range.split('-'))
            for ID_range in data[0].split('\n')
        ]
        available_IDs = [
            int(ID) for ID in data[1].split('\n')
        ]
        number_of_fresh_ingredients = 0
        for ID in available_IDs:
            for fresh_IDs in fresh_ID_ranges:
                if fresh_IDs[0] <= ID <= fresh_IDs[1]:
                    number_of_fresh_ingredients += 1
                    break
        return number_of_fresh_ingredients

    def solve_part2(self, data: Tuple[str,str]) -> int:
        fresh_ID_ranges = sorted(
            [
                list(int(endpoint) for endpoint in ID_range.split('-'))
                for ID_range in data[0].split('\n')
            ],
            key=lambda x: x[0]
        )
        i = 1
        while i < len(fresh_ID_ranges):
            if fresh_ID_ranges[i][0] <= fresh_ID_ranges[i-1][1]:
                fresh_ID_ranges[i-1][1] = max(
                    fresh_ID_ranges[i-1][1],
                    fresh_ID_ranges[i][1]
                )
                del fresh_ID_ranges[i]
            else:
                i += 1
        number_of_fresh_ingredients = sum(
            ID_range[1] - ID_range[0] + 1
            for ID_range in fresh_ID_ranges
        )
        return number_of_fresh_ingredients
```

```python
solver = Solver(PUZZLE_YEAR, PUZZLE_DAY, AoC_SESSION)
```

# Build part 1 test case(s)

```python
puzzle_instructions = solver.puzzle_instructions

part1_test_input = solver.get_value_after('For example:').create_tuple(str,'\n\n')
print(f'{part1_test_input=}\n')

part1_test_output = solver.get_value_after('So, in this example, ').as_int
print(f'{part1_test_output=}\n')
```

```python
part_1_tester = AoCTester()
part_1_tester.add_test_case(part1_test_input, part1_test_output)
```

```python
%%time
part_1_tester.run_tests(solver.solve_part1)
```

# Determine part 1 solution

```python
%%time
puzzle_input = solver.puzzle_input.create_tuple(str,'\n\n')
part1_solution = solver.solve_part1(puzzle_input)
print(f'{part1_solution=}\n')
```
